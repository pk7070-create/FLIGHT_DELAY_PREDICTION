```python
#changing the airline column into th numeric 
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

flights=pd.read_csv("data/flights.csv")
flights
encoder=OneHotEncoder(sparse_output=False,drop='first')
airline_encoded=encoder.fit_transform(flights[['AIRLINE']])
airline_df=pd.DataFrame(airline_encoded,columns=encoder.get_feature_names_out(['AIRLINE']))
flights = pd.concat(
    [flights.drop('AIRLINE', axis=1), airline_df],
    axis=1
)

flights
```

```python
flights.drop('TAIL_NUMBER', axis=1, inplace=True)
flights
```

```python
flights.duplicated().sum()
flights.drop_duplicates(inplace=True)
```

```python
#GETTING UNIQUE VALUES OF AIRPORT 
flights['ORIGIN_AIRPORT'].nunique()
```

```python
#TAKING TOP 10 AIRPORTS 
top_origins = flights['ORIGIN_AIRPORT'].value_counts().head(10).index
flights['ORIGIN_AIRPORT'] = flights['ORIGIN_AIRPORT'].apply(
    lambda x: x if x in top_origins else 'OTHER'
)
flights['ORIGIN_AIRPORT'].value_counts()
```

```python
#APPLYING ONE HOT ENCODER TO ORIGIN_AIRPORT
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
encoded1=OneHotEncoder(sparse_output=False,drop='first')
origin_encoded=encoded1.fit_transform(flights[['ORIGIN_AIRPORT']])
origin_df=pd.DataFrame(origin_encoded,columns=encoded1.get_feature_names_out(['ORIGIN_AIRPORT']),index=flights.index)
flights = pd.concat(
    [flights.drop('ORIGIN_AIRPORT', axis=1), origin_df],
    axis=1)

flights
```

```python
#DESTINATION AIRPORT
flights['DESTINATION_AIRPORT'].nunique()
```

```python
top_destinations = flights['DESTINATION_AIRPORT'].value_counts().head(10).index
flights['DESTINATION_AIRPORT'] = flights['DESTINATION_AIRPORT'].apply(
    lambda x: x if x in top_destinations else 'OTHER'
)
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

encoder = OneHotEncoder(sparse_output=False, drop='first')

dest_encoded = encoder.fit_transform(flights[['DESTINATION_AIRPORT']])

dest_df = pd.DataFrame(
    dest_encoded,
    columns=encoder.get_feature_names_out(['DESTINATION_AIRPORT']),
    index=flights.index
)

flights = pd.concat(
    [flights.drop('DESTINATION_AIRPORT', axis=1), dest_df],
    axis=1
)

flights
```

```python
#getting the target cloumns delayed.
flights['DELAYED']=flights['DEPARTURE_DELAY'].apply(lambda x:1 if x>0 else 0)
flights[['DELAYED','DEPARTURE_DELAY']].head()
flights[['DELAYED','DEPARTURE_DELAY']].tail()
```

```python
#making my data short.
flights_sample=flights.sample(100000,random_state=42)
flights_sample
```

```python
#taking x and y
x=flights_sample.drop(columns=['DELAYED','DEPARTURE_DELAY'])
y=flights_sample['DELAYED']
```

```python
#training and testing 
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(
    x,y,
    test_size=0.2,
    random_state=42
)
```

```python
#setting up Logistic Regression 
from sklearn.linear_model import LogisticRegression
model=LogisticRegression(
    max_iter=200,
    solver='saga',
    n_jobs=-1
)
```

```python
#now training model
model.fit(x_train,y_train)
```

```python
#now prediction
y_pred=model.predict(x_test)
y_pred
```

```python
#checking the accuracy store
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
#checking for confusion matrix
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
import joblib
joblib.dump(model, "flight_delay_model.pkl")
```
